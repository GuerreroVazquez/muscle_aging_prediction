```python
file_list = [
     "feature_importance_smote_catboost_7b_top_genes_elbow.csv",
    "feature_importance_smote_catboost_7b_top_genes_inflexion.csv",
    "feature_importance_smote_ridge_7b_top_genes_elbow.csv",
    "feature_importance_smote_ridge_7b_top_genes_inflexion.csv"]

path = "/home/karen/Documents/GitHub/Identify_muscle_age_genes/deep_learning/Results/feature_selection/RNAseq/"

names = ["catboost_elbow", "catboost_inflexion", "ridge_elbow", "ridge_inflexion"]

```

```python
import pandas as pd
from functools import reduce

# Define the list of CSV file paths
def read_and_process_csv(file):
    """Reads a CSV, drops the first column, and returns the Ensembl column as a set."""
    df = pd.read_csv(file)
    # Ignore the first column and keep only 'Ensembl'
    df = df.drop(df.columns[0], axis=1)  # Drop first column by index
    return set(df['Ensembl'])

# Process each file in file_list
ensembl_sets = [read_and_process_csv(path+file) for file in file_list]

# Find the intersection of all sets
intersection_all = reduce(lambda x, y: x & y, ensembl_sets)

# Find the intersection of every pair
pairwise_intersections = {}
for i in range(len(ensembl_sets)):
    for j in range(i + 1, len(ensembl_sets)):
        pairwise_intersections[f"File {names[i]} & File {names[j]}"] = ensembl_sets[i] & ensembl_sets[j]

# Print results
print("Intersection of all files:")
print(intersection_all)

print("\nPairwise intersections:")
for pair, intersect in pairwise_intersections.items():
    print(f"{pair}: {len(intersect)} common terms")
```

```python
with open('intersections_results.txt', 'w') as f:
    # Write the intersection of all files
    f.write("Intersection of all files:\n")
    f.write(f"{len(intersection_all)} common terms:\n")
    f.write("\n".join(intersection_all) + "\n\n")
    
    # Write the pairwise intersections
    f.write("Pairwise intersections:\n")
    for pair, intersect in pairwise_intersections.items():
        f.write(f"{pair}: {len(intersect)} common terms\n")
        f.write("Repeated elements:\n")
        f.write("\n".join(intersect) + "\n\n")

print("Results written to 'intersections_results.txt'")
```
