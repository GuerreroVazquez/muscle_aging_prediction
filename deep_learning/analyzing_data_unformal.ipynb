```python
import pandas as pd
```

```python
dataset_path='Data/trainign_dataset_51270_microarray_adjusted.csv'

df = pd.read_csv(dataset_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Generate some sample data
data = df

# Histogram
plt.hist(data, bins=30, density=True, alpha=0.6, color='g')

# Q-Q plot
stats.probplot(data, dist="norm", plot=plt)
plt.show()
```

```python
features = df.columns
features = features[1:-4]
```

```python
len(features)
```

```python
stats_dic = {"Shapiro-Wilk Test": {},"Kolmogorov-Smirnov Test": {}}
double_check_s= []
double_check_k = []
for feature in features:
    data = df[feature]
    # Shapiro-Wilk test
    stat, p = stats.shapiro(data)
    stats_dic["Shapiro-Wilk Test"][feature] = {"Stat": stat ,"p-value": p}
    if p < 0.1:
        double_check_s.append(feature)
        #print(f'{feature} - Shapiro-Wilk Test - Statistic:', stat, 'p-value:', p)
    # Kolmogorov-Smirnov test
    
    stat, p = stats.kstest(data, 'norm')
    stats_dic["Kolmogorov-Smirnov Test"][feature] = {"Stat": stat,"p-value": p}

    if p < 0.1:
        double_check_k.append(feature)
    #print(f'{feature} - Kolmogorov-Smirnov Test - Statistic:', stat, 'p-value:', p)
print(len(double_check_s))
print(len(double_check_k))
```
