```python
import pandas as pd
import os
```

```python

def process_csv(folder_path, file_name):
    # Read CSV file
    df = pd.read_csv(folder_path+file_name)

    df = df.T
    df.columns = df.iloc[0]
    # Drop the first row since it's now the column index
    df = df.drop(df.index[0])
    #df = df[df.index.str.startswith('abundance')]
    df["Experiment"] = file_name.split("_")[0]
    print(file_name)
    return df
```

```python
folder_name = "Data/RNAseq/"
files_in_folder = os.listdir(folder_name)
```

```python
united_df = pd.DataFrame()
for file in files_in_folder:
    if file.endswith(".csv"):
        df = process_csv(folder_name, file)
        united_df = pd.concat([united_df, df])
united_df
```
