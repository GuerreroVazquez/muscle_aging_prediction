```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind
```

```python
path_samples = 'data/test_sample_data.csv'
sample_data = pd.read_csv(path_samples )
sample_data['Diff_catboost'] = sample_data['CatboostP'] - sample_data['Age']
sample_data['Diff_ridge'] = sample_data['RidgeP'] - sample_data['Age']
```

```python
# Reshape data
melted_data = sample_data.melt(id_vars=['Status'], value_vars=['Diff_catboost', 'Diff_ridge'], 
                               var_name='Model', value_name='Difference')

# Create boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(data=melted_data, x='Status', y='Difference', hue='Model')

# Title and legend
plt.title("Age difference by ")
plt.legend(title="Model")
plt.savefig('age_prediction_difference_boxplot.svg')
plt.show()
```
